```python
import pandas as pd

df1 = pd.read_csv('2024DefenseData.csv')
df2 = pd.read_csv('2025DefenseData.csv')
df1['year'] = 2024
df2['year'] = 2025

print("File 1 Columns:", df1.columns.tolist())
print("File 2 Columns:", df2.columns.tolist())
```

```python
# Merge them side-by-side
combined_df = pd.concat([df1, df2], ignore_index=True)

# Preview the combined data right inside your notebook
combined_df.head()
```

```python
combined_df.to_csv('DefenseData.csv', index=False)
```
