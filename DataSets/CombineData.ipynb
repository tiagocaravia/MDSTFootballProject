```python
import pandas as pd

df1 = pd.read_csv('2024DefenseData.csv')
df2 = pd.read_csv('2025DefenseData.csv')
df1['year'] = 2024
df2['year'] = 2025

print("File 1 Columns:", df1.columns.tolist())
print("File 2 Columns:", df2.columns.tolist())
```

```python
# Merge them side-by-side
combined_df = pd.concat([df1, df2], ignore_index=True)

# Preview the combined data right inside your notebook
combined_df.head()
```

```python
combined_df.to_csv('DefenseData.csv', index=False)
```

```python
df3 = pd.read_csv('2024ReceivingData.csv')
df4 = pd.read_csv('2025ReceivingData.csv')
df3['year'] = 2024
df4['year'] = 2025

print("File 3 Columns:", df3.columns.tolist())
print("File 4 Columns:", df4.columns.tolist())
```

```python
dfReceive = pd.read_csv('ReceivingData.csv')
dfReceive.head()
```

```python
dfReceive = dfReceive.rename(columns={
    "Yds": "ReceiveYds",
    "TD": "ReceiveTD",
    "1D": "Receive1D",
    "Y/G": "ReceiveY/G"
})

dfReceive.head()
```

```python
dfRush = pd.read_csv('RushingData.csv')
dfRush = dfRush.rename(columns={
    "Yds": "RushYds",
    "TD": "RushTD",
    "1D": "Rush1D",
    "Att" : "RushAtt",
    "Y/G": "RushY/G"
})

dfRush.head()
```

```python
dfPass = pd.read_csv('PassingData.csv')
dfPass = dfPass.rename(columns={
    "Yds": "PassYds",
    "TD": "PassTD",
    "1D": "Pass1D",
    "Att" : "PassAtt",
    "Y/G": "PassY/G"
})

dfPass.head()
```

```python
FantasyDF = pd.concat([dfRush, dfPass, dfReceive], ignore_index=True)
FantasyDF.head()
FantasyDF.to_csv('FantasyData.csv', index=False)
```
